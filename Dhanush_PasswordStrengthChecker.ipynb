## Objective
Check a password against basic security rules and provide a strength rating and suggestions. The password is processed locally in this notebook and is not sent to a website or server.

## Rules checked
- Minimum length of 8 characters
- Uppercase and lowercase letters
- Numbers
- Special characters
- Common/weak password examples

> Educational note: This is a simple demonstration, not a complete password-auditing tool. Do not enter a real password you currently use.


```python
import re

COMMON_PASSWORDS = {
    "password", "123456", "123456789", "qwerty", "admin",
    "letmein", "welcome", "password123", "iloveyou"
}

def check_password_strength(password):
    """Return a strength rating and practical improvement suggestions."""
    score = 0
    suggestions = []

    if len(password) >= 8:
        score += 1
    else:
        suggestions.append("Use at least 8 characters; longer is better.")

    if re.search(r"[a-z]", password):
        score += 1
    else:
        suggestions.append("Add a lowercase letter.")

    if re.search(r"[A-Z]", password):
        score += 1
    else:
        suggestions.append("Add an uppercase letter.")

    if re.search(r"\d", password):
        score += 1
    else:
        suggestions.append("Add a number.")

    if re.search(r"[^A-Za-z0-9]", password):
        score += 1
    else:
        suggestions.append("Add a special character, such as ! or #.")

    if password.lower() in COMMON_PASSWORDS:
        score = 0
        suggestions.append("Avoid common passwords; choose a unique passphrase.")

    if score <= 2:
        rating = "Weak"
    elif score <= 4:
        rating = "Moderate"
    else:
        rating = "Strong"

    if not suggestions:
        suggestions.append("Good job. Consider using a longer, unique passphrase.")

    return rating, suggestions
```

```python
# Try sample passwords only. Do not use your real password.
sample_passwords = ["123456", "hello123", "Sunshine!42"]

for sample in sample_passwords:
    rating, tips = check_password_strength(sample)
    print(f"Password: {sample} | Strength: {rating}")
    for tip in tips:
        print(" -", tip)
    print()
```
